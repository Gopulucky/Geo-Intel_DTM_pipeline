```python
import os
import sys

# ==================== CONFIGURATION ====================
# 1-10: Specific Village, 11: ALL VILLAGES
VILLAGE_CHOICE = 1
# 1: DTM, 2: Hydrology, 3: Visualization, 4: ALL
PROCESS_CHOICE = 4
# =======================================================

VILLAGES = [
    "DEVDI_511671", "KHAPRETA_510206", "Dhal_Hoshiarpur_31235",
    "DHUNDA_FATEHGARH_SAHIB_32619", "67169_5NKR_CHAKHIRASINGH",
    "64334_2H_REFLIGHT", "PIRAYANKUPPAM", "THANDALAM",
    "Gandhinagar_Diglipur", "Kadamtala_Rangat"
]

def run_pipeline():
    # Validate Village Selection
    if 1 <= VILLAGE_CHOICE <= 10:
        selected_villages = [VILLAGES[VILLAGE_CHOICE-1]]
    elif VILLAGE_CHOICE == 11:
        selected_villages = VILLAGES
    else:
        print("Invalid VILLAGE_CHOICE.")
        return

    # Map Process Selection to Scripts
    scripts = []
    if PROCESS_CHOICE in [1, 4]: scripts.append("dtm_drainage_pipeline.py")
    if PROCESS_CHOICE in [2, 4]: scripts.append("colab_hydrology_pipeline.py")
    if PROCESS_CHOICE in [3, 4]: scripts.append("overlay_visualiser.py")

    if not scripts:
        print("Invalid PROCESS_CHOICE.")
        return

    for vill in selected_villages:
        for script in scripts:
            if not os.path.exists(script):
                print(f"⚠️  Warning: {script} not found in {os.getcwd()}")
                continue

            print(f"\n{'-'*50}")
            print(f"🚀 Executing {script} for {vill}...")
            print(f"{'-'*50}")

            # Setting sys.argv allows the scripts to receive the village name as an argument
            sys.argv = [script, vill]
            try:
                with open(script, 'r') as f:
                    exec(f.read(), globals())
            except Exception as e:
                print(f"❌ Error running {script}: {e}")

    print("\n" + "="*50)
    print("  PIPELINE EXECUTION COMPLETE")
    print("="*50)

if __name__ == "__main__":
    run_pipeline()
```
