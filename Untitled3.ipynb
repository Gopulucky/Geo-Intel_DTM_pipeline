```python
# 3. Discover Datasets (Villages)
import glob

# Assumes LAS/LAZ files are inside 'data' or 'input_data' within PROJECT_DIR
data_dirs = ["/content/drive/MyDrive/GEO-INTEL_pipeline/Datasets"]
las_files = []
for d in data_dirs:
    las_files.extend(glob.glob(os.path.join(d, "**", "*.la[sz]"), recursive=True))
    las_files.extend(glob.glob(os.path.join(d, "**", "*.LA[SZ]"), recursive=True))

villages = list(set([os.path.splitext(os.path.basename(f))[0] for f in las_files]))
print("Found datasets for villages:", villages)
if not villages:
    print("No .las or .laz files found! Please ensure your point cloud data is uploaded.")
```

```python
# 4. Run Pipelines for All Villages
import os
import shutil

# Configuration
PROJECT_DIR = '/content/drive/MyDrive/GEO-INTEL_pipeline'
INPUT_DIR = os.path.join(PROJECT_DIR, 'input_data')
OUTPUTS_DIR = os.path.join(PROJECT_DIR, 'outputs')

# Fix for lulc_pipeline hardcoded path error
# The script expects: /content/drive/MyDrive/model/outputs
LEGACY_PARENT = '/content/drive/MyDrive/model'
if not os.path.exists(LEGACY_PARENT):
    os.makedirs(LEGACY_PARENT, exist_ok=True)
    print(f"Created legacy parent directory: {LEGACY_PARENT}")

LEGACY_OUTPUTS = os.path.join(LEGACY_PARENT, 'outputs')
if not os.path.exists(LEGACY_OUTPUTS):
    os.symlink(OUTPUTS_DIR, LEGACY_OUTPUTS)
    print("Created symlink for lulc_pipeline legacy path.")

# Ensure primary directories exist
os.makedirs(INPUT_DIR, exist_ok=True)
os.makedirs(OUTPUTS_DIR, exist_ok=True)

os.chdir(PROJECT_DIR)
print(f"Working directory: {os.getcwd()}")

# Step 4a: Link discovered LAS/LAZ files to the expected 'input_data' folder
print("Linking datasets to input_data folder...")
for filepath in las_files:
    dest = os.path.join(INPUT_DIR, os.path.basename(filepath))
    if not os.path.exists(dest):
        try:
            os.symlink(filepath, dest)
        except:
            shutil.copy(filepath, dest)

scripts = [
    "GEO_INTEL_pipeline.py",
    "Hydrology_pipeline.py",
    "lulc_pipeline.py",
    "InteractiveMap.py"
]

for village in villages:
    print("\n" + "="*60)
    print(f"  PROCESSING VILLAGE: {village}")
    print("="*60)
    for script in scripts:
        if os.path.exists(script):
            print(f"\n---> Running {script} for {village}...")
            !python "{script}" "{village}"
        else:
            print(f"\n---> Warning: Script {script} not found in {os.getcwd()}")

print("\nAll processing complete!")
```
